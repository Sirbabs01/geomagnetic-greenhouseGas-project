## Violin plot analysis on the variables to show the relationships between variables and their relative distribution with eah other

### For Dst

```python
plt.figure(figsize=(10,8))
ax = sns.violinplot(x=Dst_year['YEAR'], y=Dst_year['DST DAILY AVERAGE'], data=Dst_year, inner='box', hue='YEAR', palette='coolwarm')
# but we want to overlap the values of the means on the plot along side as points
means = Dst_year.groupby('YEAR')['DST DAILY AVERAGE'].mean()
sns.pointplot(x=means.index, y=means.values, color='red', markers='D')

# remove the default legend in the violin plot
ax.legend_.remove()

# to get the labels and handles for the violin plot
handles, labels = ax.get_legend_handles_labels()
# create handles and labels for the mean plot
from matplotlib.lines import Line2D
mean_handle = Line2D([0], [0], marker='D', color='red', linestyle='None')

# add it to the handles and labels of the violin plot
handles.append(mean_handle);
labels.append('Yearly Mean Points')

# now assign the labels and handles to labels and handles in legend
ax.legend(handles=handles, labels=labels, frameon=False, loc='upper right', ncol=5)
# assign the x and y labels
plt.title('Violin plot of DST DAILY AVERAGE across the years')
plt.xlabel('YEAR')
plt.ylabel('DST DAILY AVERAGE')
plt.xticks(rotation=70)
plt.tight_layout()
plt.savefig('The violine plot of Dst distribution',dpi=300, bbox_inches='tight')
plt.show()
```

# Plot a time series plot for DST Index, TEC, hmF2 and foF2

## For DST Index

```python
# slice out the columns DATE, YEAR and DST DAILY AVERAGE
dst_time = df[["DATE", "DST DAILY AVERAGE", "YEAR"]]
dst_time.head()
```

```python
dst_time.dropna(inplace = True)
dst_time.head()
```

```python
# plot DST against Time
dst_time['DATE'] = pd.to_datetime(df['DATE'])
dst_time.set_index('DATE', inplace = True)
plt.figure(figsize = (12,6))
plt.plot(dst_time.index, dst_time['DST DAILY AVERAGE'])
plt.title("DST time series")
plt.xlabel("Time (seconds)")
plt.ylabel("Geomagnetic Strength (DST)")
plt.grid(True)
plt.tight_layout()
plt.savefig('Dst Time series.png', dpi=300, bbox_inches='tight')
plt.show()
```

#### Perform a stationarity test on the DST Time series

##### A Rolling statisitcs test on the timeseries

```python
# assign the DST column to ts (i.e ts for time series)
ts = dst_time['DST DAILY AVERAGE']
# Rolling Statistics
rolling_mean = ts.rolling(window = 365).mean()
rolling_std = ts.rolling(window = 365).std()

# plotting the values now
plt.figure(figsize = (12,6))
plt.plot(ts, color = 'blue', label = 'Original Data')
plt.plot(rolling_mean, color = 'red', label = 'Rolling Mean (365 Days)')
plt.plot(rolling_std, color = 'black', label = 'Rolling STD(365 DAYS)')
plt.title("Visual Stationarity test for DST Time Series")
plt.legend()
plt.grid(True)
plt.savefig('stationarityDST.png', dpi=300, bbox_inches='tight')
plt.show()
```

##### A ADCF TEST on the time series

```python
from statsmodels.tsa.stattools import adfuller

result = adfuller(ts.dropna())  # drop NA values
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])
print('Critical Values:')
for key, value in result[4].items():
    print('\t%s: %.3f' % (key, value))
```

## Plot the time series for each of these seperate years and perform the stationarity test for each

### For DST time series

```python
## Seperate the dataset into each year
# print all the unique year values
years = dst_time['YEAR'].unique()
print(years)

# create a dictionary of yearly dataframes
yearly_data = {year: dst_time[dst_time['YEAR'] == year] for year in years}
```

```python
# create a function that will plot all the timeseries and perform all the stationarity tests
def time_series(year, x, dv):
    # the parameters are year, x which is the variable we are plotting a time series and a dv variable for the title
    # assign year to a variable m
    m = year
    print(f"TIme series and stationarity test analysis for year {m}")
    # assign the dataset to year
    year = yearly_data[year]
    
    # plot the time series
    plt.figure(figsize = (12,6))
    plt.plot(year.index, year[x])
    plt.title(f"{dv} time series Plot for {m}")
    plt.xlabel("Time (seconds)")
    plt.ylabel(f"{m} {dv}")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig(f"{m} {x} Timeseries.png", dpi=300, bbox_inches='tight')
    plt.show()

    # perform the stationarity tests (Rolling Test and ADF Test)
    # assign the year column to ts (i.e ts for time series)
    ts = year[x]
    # Rolling Statistics
    rolling_mean = ts.rolling(window = 30).mean()
    rolling_std = ts.rolling(window = 30).std()

    # plotting the values now
    plt.figure(figsize = (12,6))
    plt.plot(ts, color = 'blue', label = 'Original Data')
    plt.plot(rolling_mean, color = 'red', label = 'Rolling Mean (30 Days)')
    plt.plot(rolling_std, color = 'black', label = 'Rolling STD(30 DAYS)')
    plt.title(f"Visual Stationarity test for {dv} Time Series plot for {m}")
    plt.legend()
    plt.grid(True)
    plt.show()

    from statsmodels.tsa.stattools import adfuller

    if year[x].std() == 0:
        print("Skipped: Constant Time series")
    else:
        result = adfuller(ts.dropna())  # drop NA values
        print('ADF Statistic: %f' % result[0])
        print('p-value: %f' % result[1])
        print('Critical Values:')
        for key, value in result[4].items():
            print('\t%s: %.3f' % (key, value))
            
        # Print whether plot is Stationary or non - stationary
        if (result[0] > result[4]['1%']) and (result[0] > result[4]['5%']) and (result[0] > result[4]['10%']):
            print('The time series is not stationary')
        else:
            print('The Time series is Stationary')
```

```python
# assign the parameters of the function
x = 'DST DAILY AVERAGE'
dv = 'Geomagnetic Disturbance (DST)'
# use for loop to plot the time series for each year and also check the stationarity test for the list years
for year in years:
    time_series(year, x, dv)
```

### Seperate the dst_tec dataframe into seperate datasets for each year

```python
# print all the unique year values
years = dst_tec['YEAR'].unique()
print(years)

# create a dictionary of yearly dataframes
yearly_data = {year: dst_tec[dst_tec['YEAR'] == year] for year in years}
```

### Create a function to print the first 5 rows of all the years, plot a scatter plot and find the pearson, spearman and kendall correlation

```python
# The function will be called relation_analysis
def relation_analysis(year, x, y):
    # assign the year argument to another variabe m
    m = year
    # print the first 5 rows of the dataset
    print(f"Analysis for year {year}")
    year = yearly_data[year]
    print(year.head())
    
    # plot the scatter plot 
    plt.figure(figsize = (10,6))
    plt.scatter(year[x], year[y])
    plt.title(f"A plot of DST vs TEC for {m}")
    plt.xlabel("Geomagnetic Strength (DST)")
    plt.ylabel("Total Electron Content (10^16 e/m^2)")
    plt.grid(True)
    plt.show()

    # find the pearson, spearman and kendall correlation coefficient
    p = year[x].corr(year[y], method = 'pearson')
    s = year[x].corr(year[y], method = 'spearman')
    k = year[x].corr(year[y], method = 'kendall')
    print(f"The Pearson correlation coefficient is {p}")
    print(f"The Spearman correlation coefficient is {s}")
    print(f"The Kendall correlation coefficient is {k}")
```

#### Use the function to show the scatter plot relationship between DST and TEC for every year

## Create Model

```python
# import models
from pygam import LinearGAM, s, te
from sklearn.metrics import r2_score, mean_squared_error
from sklearn.model_selection import train_test_split
```

```python
df.head()
```

```python
df.info()
```

```python
# from the plots above it is clear that solar cycles affect our dataset, so to include it as a feature in out dataset to capture seasonality
solar_cycles = {
    21: pd.Timestamp("1976-03-01"),
    22: pd.Timestamp("1986-09-01"),
    23: pd.Timestamp("1996-08-01"),
    24: pd.Timestamp("2008-12-01"),
    25: pd.Timestamp("2019-12-01")
}
```

```python
# create a function that will help assign solar cycle as a new column in my dataset
# And phase to show how long into the solar cycle has been covered based on current data
def get_solar_cycle_phase(date):
    for cycle, start in sorted(solar_cycles.items()):
        if cycle < 25:
            end = solar_cycles[cycle + 1]
        else:
            end = pd.Timestamp("2023-01-01")
            # estimated end of solar cycle 25
        if start <= date < end:
            total_days = (end - start).days
            phase = (date - start).days / total_days
            return cycle, phase
    return None, None
```

```python
## Add solar cycle id and phase as columns to my dataset
df['solar_cycle'], df['cycle_phase'] = zip(*df['DATE'].apply(get_solar_cycle_phase))
```

```python
df_model = df[["YEAR", "DATE", "DST DAILY AVERAGE", "CO2_Conc", "cycle_phase", "Average of TEC", "Average of hmF2", "Average of foF2"]]
df_model.head()
```

```python
# To test how the ionospheric paramerers respond to Dst's effect. Is it lagged or immediate
# lag the dst values by 1 day to check whether the ionospheric response to geomagnetic disturbbances is lagged by a day
df_model['Dst_lag1'] = df['DST DAILY AVERAGE'].shift(1)
# this was done now as to lag the Dst value, before redmoving missing values, so that the temporal flow is not affected 
```

#### How is output depending on the input (partial dependence plots)

```python
# How is output depending on the input (partial dependence plots)
for i, term in enumerate(gamDefault.terms):
    if term.isintercept:
        continue

    XX = gamDefault.generate_X_grid(term = i)
    pdep, confi = gamDefault.partial_dependence(term = i, X = XX, width = 0.95)

    
    
    plt.figure()
    plt.plot(XX[:, term.feature], pdep, label='Partial Dependence', color='blue')
    plt.plot(XX[:, term.feature], confi[:, 0], 'r--', label='Lower 95% CI')
    plt.plot(XX[:, term.feature], confi[:, 1], 'r--', label='Upper 95% CI')
    plt.fill_between(XX[:, term.feature], confi[:, 0], confi[:, 1], color='red', alpha=0.2)  # optional: shaded area
    plt.title(repr(term))
    plt.legend()
    plt.show()
```

### Bulid the model using the right penalisation for wiggliness and the actual number of splines that will explain the most deviance

#### Create a series of models and check the total deviance explained using different n_splines and penalization

```python
# create an array of possible splines
splines = np.arange(5, 105, 5)
# create an array of possible lambda values
lambdas = np.logspace(-4, 5, 10)

# create an empty list for splines, lambdas
spline_value = []
lambda_value = []
# create empty lists to put the value that would be used to evaluate the model produced by each spline contribution
variance = []
AIC_values = []
loglik = []
gcv = []
scale = []
edf = []
deviance = []

# use a for loop to loop through all the possible combinations of spline and lambda values to select the one with the most variance
for i in splines:
    # use a second for loop to loop through all the lambda values
    for j in lambdas:
        gam = LinearGAM(s(0, lam = j, n_splines = i) +
                        s(1, lam = j, n_splines = i) +
                        s(2, lam = j, n_splines = i)
                       ).fit(x_train, y_train)
        # calculate the full deviance for each combination
        fde = gam.statistics_['pseudo_r2']['explained_deviance']
        # calcuate the AIC score for each combination
        aic = gam.statistics_['AIC']
        # calculate how likely the model would generate my observed data
        lig = gam.statistics_['loglikelihood']
        # calculate the gcv value
        gv = gam.statistics_['GCV']
        # calculate the residual not expalined by model (i.e scale)
        sc = gam.statistics_['scale']
        # calculate the effective degrees of freedom
        ef = gam.statistics_['edof']
        # calcuate the deviance ('measure of how bad the model is')
        dv = gam.statistics_['deviance']
        # now append each of the new values to their respective lists
        spline_value.append(i)
        lambda_value.append(j)
        variance.append(fde)
        AIC_values.append(aic)
        loglik.append(lig)
        gcv.append(gv)
        scale.append(sc)
        edf.append(ef)
        deviance.append(dv)
    # now the for loop will keep repeating itself and lists will keep getting bigger to stor all the values

# After that we compile all the lists together as a dictionary and then turn it into a dataframe
results = {"spline_no" : spline_value, "lambda_term" : lambda_value, "AIC_score" : AIC_values, "Log-likelihood" : loglik, "GCV" : gcv, "Scale" : scale, "Edof" : edf, "Deviance" : deviance, "Full Variance Explained" : variance}
# now convert it to a dataframe
df_results = pd.DataFrame(results)
df_results
```

```python
# check the values of the possible R^2 for the results
df_results["Full Variance Explained"].unique()
```

### Build a PyGAM model for foF2 using the appropriate no of splines and lambda values

#### Create a series of models with different combinatins of no of splines and lambda values for predicting foF2 

```python
# create an array of possible splines
splines = np.arange(5, 105, 5)
# create an array of possible lambda values
lambdas = np.logspace(-4, 5, 10)

# create an empty list for splines, lambdas
spline_value = []
lambda_value = []
# create empty lists to put the value that would be used to evaluate the model produced by each spline contribution and lambda contribution
variance = []
AIC_values = []
loglik = []
gcv = []
scale = []
edf = []
deviance = []

# use a for loop to loop through all the possible combinations of spline and lambda values to select the one with the most variance
for i in splines:
    # use a second for loop to loop through all the lambda values
    for j in lambdas:
        gam = LinearGAM(s(0, lam = j, n_splines = i) +
                        s(1, lam = j, n_splines = i) +
                        s(2, lam = j, n_splines = i)
                       ).fit(x2_train, y2_train)
        # calculate the full deviance for each combination
        fde = gam.statistics_['pseudo_r2']['explained_deviance']
        # calcuate the AIC score for each combination
        aic = gam.statistics_['AIC']
        # calculate how likely the model would generate my observed data
        lig = gam.statistics_['loglikelihood']
        # calculate the gcv value
        gv = gam.statistics_['GCV']
        # calculate the residual not expalined by model (i.e scale)
        sc = gam.statistics_['scale']
        # calculate the effective degrees of freedom
        ef = gam.statistics_['edof']
        # calcuate the deviance ('measure of how bad the model is')
        dv = gam.statistics_['deviance']
        # now append each of the new values to their respective lists
        spline_value.append(i)
        lambda_value.append(j)
        variance.append(fde)
        AIC_values.append(aic)
        loglik.append(lig)
        gcv.append(gv)
        scale.append(sc)
        edf.append(ef)
        deviance.append(dv)
    # now the for loop will keep repeating itself and lists will keep getting bigger to stor all the values

# After that we compile all the lists together as a dictionary and then turn it into a dataframe
results = {"spline_no" : spline_value, "lambda_term" : lambda_value, "AIC_score" : AIC_values, "Log-likelihood" : loglik, "GCV" : gcv, "Scale" : scale, "Edof" : edf, "Deviance" : deviance, "Full Variance Explained" : variance}
# now convert it to a dataframe
df_results2 = pd.DataFrame(results)
df_results2
```

```python
# check the values of the possible R^2 for the results
df_results2["Full Variance Explained"].unique()
```

```python
# now we rank these by the full variance explained
ranked_results2 = df_results2.sort_values(by = 'Full Variance Explained', ascending = False)
ranked_results2
```

```python
# now rank based on AIC score (remember the lower the AIC score the better the model)
rank_aic = df_results2.sort_values(by = 'AIC_score', ascending = True)
rank_aic
```

```python
# rank based on GCV (remember the lower the GCV the better the model)
rank_gcv = df_results2.sort_values(by = 'GCV', ascending = True)
rank_gcv
```

```python
# rank based on loglikelihood (remember the higher the loglikelihood the better the model)
rank_loglik = df_results2.sort_values(by = 'Log-likelihood', ascending = False)
rank_loglik
```

```python
# rank based on GCV and AIC
rank_gcvAIC = df_results2.sort_values(by = ['AIC_score', 'GCV'], ascending = [True, True])
rank_gcvAIC
```

```python
# assign the actual column names to x and y
# x for the independent variables and y for the dependent variable 
# our independent variables are DATE, DST DAILY AVERAGE, CO2_Conc
x3 = df_model[["cycle_phase", "DST DAILY AVERAGE", "CO2_Conc"]].values
y3 = df_model["Average of hmF2"].values

# in order to be able to generate the featurenames whn we are performing analysis
df_predictor = df_model[["cycle_phase", "DST DAILY AVERAGE", "CO2_Conc"]]
```

```python
# check the shape of x3
x3.shape
```

```python
# check the shape of y2
y3.shape
```

#### Split the dataset into 2, 70% for train and 30%

```python
# Split the dataset into Train and Test Sets (70% Train)
x3_train, x3_test, y3_train, y3_test = train_test_split(x3,y3, train_size = 0.7, random_state = 42)
```

##### Build and fit the model

```python
# Fit our model with all the splines being default values for dst only (feature 3 here)
gam3Default = LinearGAM(s(0) + s(1) + s(2)).fit(x3_train, y3_train)
```

```python
gam3Default.summary()
```

##### Make predictions based on training dataset

```python
# use the predict method to predict the new response variable vale
y3_pred_train = gam3Default.predict(x3_train)
```

##### Compare the predicted value againt y3_train

```python
plt.plot(y3_train, color = 'black')
plt.plot(y3_pred_train, color = 'blue')
plt.grid(True)
plt.tight_layout()
plt.show()
```

### How is the output depending on the input (partial dependence)

```python
for i, term in enumerate(gam3Default.terms):
    if term.isintercept:
        continue

    XX = gam3Default.generate_X_grid(term = i)
    pdep, confi = gam3Default.partial_dependence(term = i, X = XX, width = 0.95)

    
    
    plt.figure()
    plt.plot(XX[:, term.feature], pdep, label='Partial Dependence', color='blue')
    plt.plot(XX[:, term.feature], confi[:, 0], 'r--', label='Lower 95% CI')
    plt.plot(XX[:, term.feature], confi[:, 1], 'r--', label='Upper 95% CI')
    plt.fill_between(XX[:, term.feature], confi[:, 0], confi[:, 1], color='red', alpha=0.2)  # optional: shaded area
    plt.title(repr(term))
    plt.legend()
    plt.show()
```

```python
# so we make our model, perform a grid search out of a number of possible values to find out the best
# first we bring out our possible lambda and spline values as used in previous models
print(f"The lambda values are: {lams}")
print(f"The no of splines are: {splines}")

# up to the grid search
# Build the model but use a gridsearch for all the possible lambda and spline values
auto_model_lag3 = LinearGAM(s(0) + s(1) + s(2)).gridsearch(x_trainlag3, y_trainlag3,
    lam = lams,
    n_splines = splines
)


# Check its performance
print("Auto model R²:", auto_model_lag3.statistics_['pseudo_r2']['explained_deviance'])
print("AIC:", auto_model_lag3.statistics_['AIC'])
print("EDF:", auto_model_lag3.statistics_['edof'])
print("GCV:", auto_model_lag3.statistics_['GCV'])
print("loglikelihood:", auto_model_lag3.statistics_['loglikelihood'])
print("Deviance:", auto_model_lag3.statistics_['deviance'])
print("Scale:", auto_model_lag3.statistics_['scale'])
```

#### Plot the partial dependence plots of this model

```python
for i, term in enumerate(auto_model_lag3.terms):
    if term.isintercept:
        continue

    # Generate grid and partial dependence results
    XX = auto_model_lag3.generate_X_grid(term=i)
    pdep, confi = auto_model_lag3.partial_dependence(term=i, X=XX, width=0.95)

    # Get feature name for title
    feature_name = feature_names[term.feature]

    # Create plot
    plt.figure(figsize=(10, 5))
    plt.plot(XX[:, term.feature], pdep, label='Partial Dependence', color='blue')
    plt.plot(XX[:, term.feature], confi[:, 0], 'r--', label='Lower 95% CI')
    plt.plot(XX[:, term.feature], confi[:, 1], 'r--', label='Upper 95% CI')
    plt.fill_between(XX[:, term.feature], confi[:, 0], confi[:, 1], color='red', alpha=0.2)

    # Plot formatting
    plt.title(f"hmF2 Partial Dependence: {feature_name}", fontsize=14)
    plt.xlabel(feature_name)
    plt.ylabel("Predicted Value")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.savefig(f'pdplag3_{feature_name}.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
# so we make our model, perform a grid search out of a number of possible values to find out the best
# first we bring out our possible lambda and spline values as used in previous models
print(f"The lambda values are: {lams}")
print(f"The no of splines are: {splines}")

# up to the grid search
# Build the model but use a gridsearch for all the possible lambda and spline values
auto2_model_50 = LinearGAM(s(0) + s(1) + s(2)).gridsearch(x2_train50, y2_train50,
    lam = lams,
    n_splines = splines
)


# Check its performance
print("Auto model R²:", auto2_model_50.statistics_['pseudo_r2']['explained_deviance'])
print("AIC:", auto2_model_50.statistics_['AIC'])
print("EDF:", auto2_model_50.statistics_['edof'])
print("GCV:", auto2_model_50.statistics_['GCV'])
print("loglikelihood:", auto2_model_50.statistics_['loglikelihood'])
print("Deviance:", auto2_model_50.statistics_['deviance'])
print("Scale:", auto2_model_50.statistics_['scale'])
```

## plot the partial dependence plots

```python
for i, term in enumerate(auto2_model_50.terms):
    if term.isintercept:
        continue

    # Generate grid and partial dependence results
    XX = auto2_model_50.generate_X_grid(term=i)
    pdep, confi = auto2_model_50.partial_dependence(term=i, X=XX, width=0.95)

    # Get feature name for title
    feature_name = feature_names[term.feature]

    # Create plot
    plt.figure(figsize=(10, 5))
    plt.plot(XX[:, term.feature], pdep, label='Partial Dependence', color='blue')
    plt.plot(XX[:, term.feature], confi[:, 0], 'r--', label='Lower 95% CI')
    plt.plot(XX[:, term.feature], confi[:, 1], 'r--', label='Upper 95% CI')
    plt.fill_between(XX[:, term.feature], confi[:, 0], confi[:, 1], color='red', alpha=0.2)

    # Plot formatting
    plt.title(f"foF2 Partial Dependence for Dst < -50: {feature_name}", fontsize=14)
    plt.xlabel(feature_name)
    plt.ylabel("Predicted Value")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.savefig(f'pdp2_{feature_name}50.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.plot(y3_train50, color='black', label='Actual Values')
plt.plot(y3_pred_train_50, color='blue', label='Predicted Values')
plt.xlabel('Index')
plt.ylabel('F2-layer Peak Height (Km)')
plt.title('Model Predictions of hmF2 vs Actual hmF2 Values (Training Data)')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig('comparisonplothmF2_50(Complex).png', dpi=300, bbox_inches='tight')
plt.show()
```

## Plot the partial dependence plots

```python
for i, term in enumerate(gam3_50.terms):
    if term.isintercept:
        continue

    # Generate grid and partial dependence results
    XX = gam3_50.generate_X_grid(term=i)
    pdep, confi = gam3_50.partial_dependence(term=i, X=XX, width=0.95)

    # Get feature name for title
    feature_name = feature_names[term.feature]

    # Create plot
    plt.figure(figsize=(10, 5))
    plt.plot(XX[:, term.feature], pdep, label='Partial Dependence', color='blue')
    plt.plot(XX[:, term.feature], confi[:, 0], 'r--', label='Lower 95% CI')
    plt.plot(XX[:, term.feature], confi[:, 1], 'r--', label='Upper 95% CI')
    plt.fill_between(XX[:, term.feature], confi[:, 0], confi[:, 1], color='red', alpha=0.2)

    # Plot formatting
    plt.title(f"hmF2 Partial Dependence for Dst < -50(Complex): {feature_name}", fontsize=14)
    plt.xlabel(feature_name)
    plt.ylabel("Predicted Value")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.savefig(f'pdp3_{feature_name}50(complex).png', dpi=300, bbox_inches='tight')
    plt.show()
```

### For AP Index

```python
# select the appropriate columns
ap_time = df[['AP-Index','YEAR','DATE',]]
ap_time
```

```python
# remove missing values
ap_time.dropna(inplace = True)
ap_time
```

```python
# plot DST against Time
ap_time['DATE'] = pd.to_datetime(df['DATE'])
ap_time.set_index('DATE', inplace = True)
plt.figure(figsize = (12,6))
plt.plot(ap_time.index, ap_time['AP-Index'])
plt.title("AP Index time series")
plt.xlabel("Time (seconds)")
plt.ylabel("Geomagnetic Activity on Earth (nT)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

#### Stationarity test for time series

##### Rolling statistics test

```python
# assign the DST column to ts (i.e ts for time series)
ts = ap_time['AP-Index']
# Rolling Statistics
rolling_mean = ts.rolling(window = 365).mean()
rolling_std = ts.rolling(window = 365).std()

# plotting the values now
plt.figure(figsize = (12,6))
plt.plot(ts, color = 'blue', label = 'Original Data')
plt.plot(rolling_mean, color = 'red', label = 'Rolling Mean (365 Days)')
plt.plot(rolling_std, color = 'black', label = 'Rolling STD(365 DAYS)')
plt.title("Visual Stationarity test for AP Index Time Series")
plt.legend()
plt.grid(True)
plt.show()
```

##### ACDF test

```python
from statsmodels.tsa.stattools import adfuller

result = adfuller(ts.dropna())  # drop NA values
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])
print('Critical Values:')
for key, value in result[4].items():
    print('\t%s: %.3f' % (key, value))
```

The ADF statistics is far less than the critical values and the p-value is zero, so we can reject the null hypothesis that it is non - stationary. Therefore the time series is stationary 

## AP - Index Violin plots

```python
plt.figure(figsize=(10,8))
ax = sns.violinplot(x=ap_time['YEAR'], y=ap_time['AP-Index'], data=ap_time, inner='box', hue='YEAR', palette='coolwarm')
# but we want to overlap the values of the means on the plot along side as points
means = ap_time.groupby('YEAR')['AP-Index'].mean()
sns.pointplot(x=means.index, y=means.values, color='red', markers='D')

# remove the default legend in the violin plot
ax.legend_.remove()

# to get the labels and handles for the violin plot
handles, labels = ax.get_legend_handles_labels()
# create handles and labels for the mean plot
from matplotlib.lines import Line2D
mean_handle = Line2D([0], [0], marker='D', color='red', linestyle='None')

# add it to the handles and labels of the violin plot
handles.append(mean_handle);
labels.append('Yearly Mean Points')

# now assign the labels and handles to labels and handles in legend
ax.legend(handles=handles, labels=labels, frameon=False, loc='upper right', ncol=5)
# assign the x and y labels
plt.title('Violin plot of Daily AP-Index values across the years')
plt.xlabel('YEAR')
plt.ylabel('AP-Index')
plt.xticks(rotation=70)
plt.tight_layout()
plt.savefig('The violine plot of AP-Index distribution',dpi=300, bbox_inches='tight')
plt.show()
```

### For Bx

#### Timeseries

```python
# filter the required columns
Bx_time = df_IMF[['BX(GSE, GSM)','YEAR','DATE']]
Bx_time
```

```python
Bx_time.dropna(inplace = True)
Bx_time
```

```python
# plot DST against Time
Bx_time['DATE'] = pd.to_datetime(Bx_time['DATE'])
Bx_time.set_index('DATE', inplace = True)
plt.figure(figsize = (12,6))
plt.plot(Bx_time.index, Bx_time['BX(GSE, GSM)'])
plt.title("X-component of the Interplanetary Magnetic Field time series")
plt.xlabel("Time (seconds)")
plt.ylabel("Bx index (nT)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
#stationarity test
# assign the DST column to ts (i.e ts for time series)
ts = Bx_time['BX(GSE, GSM)']
# Rolling Statistics
rolling_mean = ts.rolling(window = 365).mean()
rolling_std = ts.rolling(window = 365).std()

# plotting the values now
plt.figure(figsize = (12,6))
plt.plot(ts, color = 'blue', label = 'Original Data')
plt.plot(rolling_mean, color = 'red', label = 'Rolling Mean (365 Days)')
plt.plot(rolling_std, color = 'black', label = 'Rolling STD(365 DAYS)')
plt.title("Visual Stationarity test for X-component Interplanetary Magnetic Field Time Series")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

result = adfuller(ts.dropna())  # drop NA values
print('ADF Statistic: %f' % result[0])
print('p-value: %f' % result[1])
print('Critical Values:')
for key, value in result[4].items():
    print('\t%s: %.3f' % (key, value))
```

## Build a model using AP as the gemagnetic parameter affecting the ionosphere

```python
# select the required coloumns
df_modelAP = df[['YEAR','DATE','cycle_phase','AP-Index','Average of TEC','Average of foF2','Average of hmF2','CO2_Conc']]
df_modelAP
```

```python
df_modelAP.dropna(inplace = True)
df_modelAP
```

```python
df_modelAP['AP-Index'].unique()
```

```python
df_modelAP['AP-Index'].value_counts()
```

### Build the model for TEC

```python
# Assign the independent variables to x and the dependent variable to y
x = df_modelAP[["cycle_phase", "AP-Index", "CO2_Conc"]].values
y = df_modelAP["Average of TEC"].values
# We already know that co2 and solar cycle phase are goood parameters, the reason for this new parameter,
# is to see how geomagnetic storms really affect the ionosphere, since Dst was not a very goood parameter
```

```python
# check the number of rows and columns of independent varaible
x.shape
```

```python
# check the number of rows and columns of dependent varaible
y.shape
```

```python
# divide the data into test and train
x_trainAP, x_testAP, y_trainAP, y_testAP = train_test_split(x,y, train_size = 0.7, random_state = 42)
```

```python
print(f"The possible values of lambda are {lams}")
print(f"The possible no of splines are {splines}")
```

```python
gamAP = LinearGAM(s(0) + s(1) + s(2)).gridsearch(x_trainAP, y_trainAP,
    lam = lams,
    n_splines = splines
)


# Check its performance
print("Auto model R²:", gamAP.statistics_['pseudo_r2']['explained_deviance'])
print("AIC:", gamAP.statistics_['AIC'])
print("EDF:", gamAP.statistics_['edof'])
print("GCV:", gamAP.statistics_['GCV'])
print("loglikelihood:", gamAP.statistics_['loglikelihood'])
print("Deviance:", gamAP.statistics_['deviance'])
print("Scale:", gamAP.statistics_['scale'])
```

```python
# Assign the performance parameters to variable names
pseudo_r2 = gamAP.statistics_['pseudo_r2']['explained_deviance']
AIC_score = gamAP.statistics_['AIC']
EDF_score = gamAP.statistics_['edof']
GCV_score = gamAP.statistics_['GCV']
loglik = gamAP.statistics_['loglikelihood']
deviance = gamAP.statistics_['deviance']
scale = gamAP.statistics_['scale']
```

```python
# check the summary of th model
gamAP.summary()
```

#### Make predictions using this model

```python
y_pred_trainAP = gamAP.predict(x_trainAP)
```

### Build the model for foF2

```python
# Assign the independent variables to x and the dependent variable to y
x = df_modelAP[["cycle_phase", "AP-Index", "CO2_Conc"]].values
y = df_modelAP["Average of foF2"].values
# We already know that co2 and solar cycle phase are goood parameters, the reason for this new parameter,
# is to see how geomagnetic storms really affect the ionosphere, since Dst was not a very goood parameter
```

```python
# divide the data into test and train
x_trainAP2, x_testAP2, y_trainAP2, y_testAP2 = train_test_split(x,y, train_size = 0.7, random_state = 42)
```

```python
# to build the best model, we have to perform a grid search over a compreensive list of lambda and no of spline values.
# lambdas to be used
lams = np.linspace(0.0001, 0.01, 20).tolist() + np.linspace(0.01, 0.1, 10).tolist()
# no of splines values o be used 
splines = list(range(5, 101, 5))
```

```python
# to build the best model we performa grid search 
gamAP2 = LinearGAM(s(0) + s(1) + s(2)).gridsearch(x_trainAP2, y_trainAP2,
    lam = lams,
    n_splines = splines
)


# Check its performance
print("Auto model R²:", gamAP2.statistics_['pseudo_r2']['explained_deviance'])
print("AIC:", gamAP2.statistics_['AIC'])
print("EDF:", gamAP2.statistics_['edof'])
print("GCV:", gamAP2.statistics_['GCV'])
print("loglikelihood:", gamAP2.statistics_['loglikelihood'])
print("Deviance:", gamAP2.statistics_['deviance'])
print("Scale:", gamAP2.statistics_['scale'])
#  here again we will be using the same lams and splines, that provides a vast number of combinations of
# lambda and no of splines (i.e. 600 combinations)
```

```python
gamAP2.summary()
```

```python
# Assign the performance parameters to variable names
pseudo_r2 = gamAP2.statistics_['pseudo_r2']['explained_deviance']
AIC_score = gamAP2.statistics_['AIC']
EDF_score = gamAP2.statistics_['edof']
GCV_score = gamAP2.statistics_['GCV']
loglik = gamAP2.statistics_['loglikelihood']
deviance = gamAP2.statistics_['deviance']
scale = gamAP2.statistics_['scale']
```

```python
# assign the predictors to x and the target to y
x = df_modelKP_storm[['cycle_phase', 'KP-Index', 'CO2_Conc']].values
y = df_modelKP_storm['Average of TEC'].values

# thanks for the predictors
df_predictor = df_modelKP[['cycle_phase', 'KP-Index', 'CO2_Conc']]
feature_names
x.shape, y.shape
```

### Split the dataset

```python
# split the dat into train test and split.
x_trainkp50, x_testkp50, y_trainkp50, y_testkp50 = train_test_split(x,y, train_size=0.7, test_size=0.3, random_state=42)
x_trainkp50.shape, y_trainkp50.shape, x_testkp50.shape, y_testkp50.shape
```

### Create list of lambda and splines values

```python
# lambdas to be used
lams = np.linspace(0.0001, 0.01, 20).tolist() + np.linspace(0.01, 0.1, 10).tolist()
# no of splines values o be used 
splines = list(range(5, 101, 5))

# Print the values
print(f"The lambda values are {lams}")
print(f"The no of spline values are {splines}")
```

### Build the model

```python
# build the model
gamkp50 = LinearGAM(s(0) + s(1) + s(2)).gridsearch(x_trainkp50, y_trainkp50, lam=lams, n_splines=splines)
gamkp50
```

### Assing the model's parameters to variables

```python
r2_scorekp50 = gamkp50.statistics_['pseudo_r2']['explained_deviance']
aickp50 = gamkp50.statistics_['AIC']
gcvkp50 = gamkp50.statistics_['GCV']
deviancekp50 = gamkp50.statistics_['deviance']
deviancekp50
```

### Check the model summary

```python
gamkp50.summary()
```

```python
# Get the feature names
feature_names = df_predictor.columns
feature_names
```

### predictions on the training data

```python
# make predictions with the model
y_pred_trainkp50 = gamkp50.predict(x_trainkp50)
```

### PLot of the comparison plots of the predicted values to the actual values

```python
plt.plot(y_trainAP3, color = 'black', label = 'Actual Value')
plt.plot(y_pred_trainAP3, color = 'blue', label = 'Predicted value')
plt.grid(True)
plt.legend()
plt.title('The comparison plot of The actual TEC values and the predicted TEC values')
plt.xlabel('The Index')
plt.ylabel('The TEC VALUES')
plt.tight_layout()
plt.savefig(f"The comparison plot of kpmodel1", dpi=300, bbox_inches='tight')
plt.show()
```

### Partial dependence plots 

```python
# plot the partial dependence plots
for i, term in enumerate(gamkp50.terms):
    if term.isintercept:
        continue

    # Generate grid and partial dependence results
    XX = gamkp50.generate_X_grid(term=i)
    pdep, confi = gamkp50.partial_dependence(term=i, X=XX, width=0.95)

    # Get feature name for title
    feature_name = feature_names[term.feature]

    # Create plot
    plt.figure(figsize=(8, 5))
    plt.plot(XX[:, term.feature], pdep, label='Partial Dependence', color='blue')
    plt.plot(XX[:, term.feature], confi[:, 0], 'r--', label='Lower 95% CI')
    plt.plot(XX[:, term.feature], confi[:, 1], 'r--', label='Upper 95% CI')
    plt.fill_between(XX[:, term.feature], confi[:, 0], confi[:, 1], color='red', alpha=0.2)

    # Plot formatting
    plt.title(f"Partial Dependence: {feature_name}", fontsize=14)
    plt.xlabel(feature_name)
    plt.ylabel("Predicted Value")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.savefig(f'pdp_{feature_name}kp50.png', dpi=300, bbox_inches='tight')
    plt.show()
```

### 4D plot of the independent varaibles are on the dependent variable

```python
# plot a 4D plot showing how the independent variables affet the dependent variables 
# import the required library for this
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(11,8))
ax = fig.add_subplot(111, projection='3d')
sc = ax.scatter(x_trainkp50[:,0], x_trainkp50[:,1], x_trainkp50[:,2], c=y_pred_trainkp50, cmap='viridis', s=40)
plt.colorbar(sc, label='Predicted TEC values')
ax.set_xlabel('Solar Cycle Phase')
ax.set_ylabel('Goemagnetic Activity Index (KP)')
ax.set_zlabel('CO2 Concentration (ppm)')
plt.title('Plot of the 3 independent variables effect on the models predictions')
plt.savefig('4D plotkp50.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Check for Concurvity in the model parameters

# Study the Interaction term between geomagnetic storms and greenhouse gases on TEC

## Assign the model parameters to variables

```python
x = df_model[['cycle_phase', 'DST DAILY AVERAGE', 'CO2_Conc']].values
y = df_model['Average of TEC'].values
x.shape, y.shape
```

## Split the dataset into train and test data

```python
x_trainGIco2, x_testGIco2, y_trainGIco2, y_testGIco2 = train_test_split(x,y, test_size=0.3, train_size=0.7, random_state=42)
x_trainGIco2.shape, y_trainGIco2.shape, x_testGIco2.shape, y_testGIco2.shape
```

```python
# import a required library
from sklearn.model_selection import ParameterSampler
```

```python
gam_GI_co2 = LinearGAM(s(0) + s(1) + s(2) + te(1,2))
```

```python
# create a hyperparameter to tune the midel for best results
param_grid = {'lam': np.logspace(-3,3,20), 'n_splines': [5,10,15,20,25,30,35,40]}
```

```python
# instead of testing all the number of splines and lambda's we randomly sample out 45 of them
sampler = list(ParameterSampler(param_grid, n_iter=45, random_state=42))
```

```python
# create variable placeholders to keep track of the best pyGAM models
best_score = -np.inf
best_gam = None

# use for loop to iterate through slected combinations
for params in sampler:
    gam = LinearGAM(s(0, n_splines=params['n_splines']) +  s(1, n_splines=params['n_splines']) + s(2, n_splines=params['n_splines']) + te(1,2, n_splines=params['n_splines']))
    gam.gridsearch(x_trainGIco2, y_trainGIco2, lam=[[params['lam']], [params['n_splines']], [params['n_splines']], [params['lam']], [params['lam']]])
    score = gam.statistics_['pseudo_r2']['explained_deviance']
    if score > best_score:
        best_score = score
        best_gam = gam
```

```python
best_gam.summary()
```

## Check the interactions of Dst and Co2 on foF2

```python
# elect the reqired columns
x = df_model[['cycle_phase', 'DST DAILY AVERAGE', 'CO2_Conc']].values
y = df_model['Average of foF2'].values
x.shape, y.shape
```

## Split the data into test and train

```python
x2_trainGIco2, x2_testGIco2, y2_trainGIco2, y2_testGIco2 = train_test_split(x,y, test_size=0.3, train_size=0.7, random_state=42)
x2_trainGIco2.shape, y2_trainGIco2.shape, x2_testGIco2.shape, y2_testGIco2.shape
```

## We want to randomly select out of the amount of possible lambda and spline combinations 45 to use for our grid search

```python
# create a hyperparameter to tune the midel for best results
param_grid2 = {'lam': np.logspace(-3,3,20), 'n_splines': [5,10,15,20,25,30,35,40]}
```

```python
# instead of testing all the number of splines and lambda's we randomly sample out 45 of them
sampler2 = list(ParameterSampler(param_grid2, n_iter=45, random_state=42))
```

## From the randomly selected combinations we perform a grid search to chose the right GAM

```python
# We keep assigning each of the best pyGAm's and the scores to variables below, for now we start with the basic zero and no gam
best_score2 = -np.inf
best_gam2 = None

# use for loop to iterate through slected combinations
for params in sampler2:
    gam2 = LinearGAM(s(0, n_splines=params['n_splines']) + s(1, n_splines = params['n_splines']) + s(2, n_splines = params['n_splines']) + te(1,2, n_splines=params['n_splines']))
    gam2.gridsearch(x2_trainGIco2, y2_trainGIco2, lam=[[params['lam']], params['lam'], params['lam'], [params['lam']], [params['lam']]])
    score = gam2.statistics_['pseudo_r2']['explained_deviance']
    if score > best_score2:
        best_score2 = score
        best_gam2 = gam2
```

## Check the interaction of Dst and co2 on hmF2

```python
# select the required columns
x = df_model[["cycle_phase", "DST DAILY AVERAGE", "CO2_Conc"]]
y = df_model["Average of hmF2"]
x.shape, y.shape
```

## Split the data into train, test data

```python
x_trainDC3, x_testDC3, y_trainDC3, y_testDC3 = train_test_split(x,y, test_size = 0.3, train_size = 0.7, random_state = 42)
x_trainDC3.shape, y_trainDC3.shape, x_testDC3.shape, y_testDC3.shape
```

## Create a sampler to randomly select 45 out of the many combinations to find the best model instead of a rigorous grid search

```python
# create a dictionary for the possible lambda and and number of spline values
lam_spline = {'lam' : np.logspace(-3,3,20), 'n_splines' : [5,10, 15, 20, 25, 30, 35, 40]}
lam_spline
```

```python
# use a parameter sampler to sample out randomly 45 models
sample3 = ParameterSampler(lam_spline, n_iter = 45, random_state = 42)
# convert the samples into a list
sample3 = list(sample3)
sample3
```

## Now perform  grid search on the randomly selected combinations of lambda and spline to choose the best model

```python
# we use a for loop where we iterate through each combintion of lam-spline to get the best combination for a generalised gam
## first we define variables where w e will store the gam score and the gam itself
best_score = -np.inf
best_gam3 = None

# now the for loop
for lam_spline in sample3:
    gam3 = LinearGAM(s(0, n_splines = lam_spline['n_splines']) + s(1, n_splines = lam_spline['n_splines']) + s(2, n_splines = lam_spline['n_splines']) + te(1,2, n_splines = lam_spline['n_splines']))
    gam3.gridsearch(x_trainDC3.values, y_trainDC3.values, lam = [[lam_spline['lam']], [lam_spline['lam']], [lam_spline['lam']], [lam_spline['lam']], [lam_spline['lam']]])
    score = gam3.statistics_['pseudo_r2']['explained_deviance']
    if score > best_score:
        best_score = score
        best_gam3 = gam3
```

## Check the best models parameters

```python
best_gam3.summary()
```

## Once again we need to check between Dst and Co2 which one is the modifier (i.e. which affects which's effect and hyow does that happen)

### First case scenario is CO2 the modifier in this interaction

```python
# so we check for Different levels of co2 how does it affect Dst's effect on the ionosphere
for co2_level in [Co2_vals.min(), np.percentile(Co2_vals, 25), np.median(Co2_vals), np.percentile(Co2_vals, 75), Co2_vals.max()]:
    # create our array of independent variables
    x_slice = np.column_stack((
        # first we want all zeros for the solar cycle phase values
        np.zeros_like(Dst_vals), # this returns an array of zero's with the same shape of the array you assign to it
        Dst_vals, # since we want the Dst values
        # now we want an array with just one co2 level the same shape as the Dst_vals array
        np.full_like(Dst_vals, co2_level) # so we get an array of only one co2 level
    ))
    # Now we predict the values of hmF2 from the x_slice
    y_pred = best_gam3.predict(x_slice)
    # now plot the Dst values against the prdicted values
    plt.plot(Dst_vals, y_pred, label = f"Co2 Level {co2_level}")
    
    
plt.xlabel('The Dst values')
plt.ylabel('The predicted hmF2 values')
plt.legend()
plt.show()
```

```python
# so we check for Different levels of co2 how does it affect Dst's effect on the ionosphere
for co2_level in [Co2_vals.min(), np.percentile(Co2_vals, 25), np.median(Co2_vals), np.percentile(Co2_vals, 75), Co2_vals.max()]:
    # create our array of independent variables
    x_slice = np.column_stack((
        # first we want all zeros for the solar cycle phase values
        np.zeros_like(kp_vals), # this returns an array of zero's with the same shape of the array you assign to it
        kp_vals, # since we want the kp values
        # now we want an array with just one co2 level the same shape as the kp_vals array
        np.full_like(kp_vals, co2_level) # so we get an array of only one co2 level
    ))
    # Now we predict the values of TEC from the x_slice
    y_pred = best_gamkp.predict(x_slice)
    # now plot the TEC values against the prdicted values
    plt.plot(kp_vals, y_pred, label = f"Co2 Level {co2_level}")
    
    
plt.xlabel('The KP values')
plt.ylabel('The predicted TEC values')
plt.title('Is CO2 modifying the geomagnetic activity effect on TEC')
plt.legend()
plt.show()
```

## Check the effect of the interaction of KP and Co2 on foF2

```python
# assign the independent variables to x and the trget to y
x = df_kp[['cycle_phase', 'KP-Index', 'CO2_Conc']]
y = df_kp['Average of foF2']

# check the shape of the x and y values
x.shape,  y.shape
```

```python
# split the data into train and test data
x_train2_kpco2, x_test2_kpco2, y_train2_kpco2, y_test2_kpco2 = train_test_split(x,y, test_size = 0.3, train_size = 0.7, random_state = 42)
x_train2_kpco2.shape, y_train2_kpco2.shape, x_test2_kpco2.shape, y_test2_kpco2.shape
```

```python
# create a dictionary for the possible lambda and and number of spline values
lam_spline_kp = {'lam' : np.logspace(-3,3,20), 'n_splines' : [5,10, 15, 20, 25, 30, 35, 40]}
lam_spline_kp
```

```python
# use a parameter sampler to sample out randomly 45 models
samplekp = ParameterSampler(lam_spline_kp, n_iter = 45, random_state = 20)
# convert the samples into a list
samplekp = list(samplekp)
samplekp
```

```python
# we use a for loop where we iterate through each combintion of lam-spline to get the best combination for a generalised gam
## first we define variables where w e will store the gam score and the gam itself
best_scorekp2 = -np.inf
best_gamkp2 = None

# now the for loop
for lam_spline in samplekp:
    gamkp2 = LinearGAM(s(0, n_splines = lam_spline['n_splines']) + s(1, n_splines = lam_spline['n_splines']) + s(2, n_splines = lam_spline['n_splines']) + te(1,2, n_splines = lam_spline['n_splines']))
    gamkp2.gridsearch(x_train_kpco2.values, y_train_kpco2.values, lam = [[lam_spline['lam']], [lam_spline['lam']], [lam_spline['lam']], [lam_spline['lam']], [lam_spline['lam']]])
    score = gamkp2.statistics_['pseudo_r2']['explained_deviance']
    if score > best_scorekp2:
        best_scorekp2 = score
        best_gamkp2 = gamkp2
```

```python
best_gamkp2.summary()
```

## We are not epecting an interaction term between Bx and co2 so e will compare some statistics test reults with that of Dst and co2

### create a function to check the range, standard deviation and mean absolute value of the interaction terms

```python
def compute_Z_and_stats(gam, term_index):
    # 1) generate grid for that term (term_index = integer index of the te term)
    XX = gam.generate_X_grid(term=term_index)
    Z_flat = gam.partial_dependence(term=term_index, X=XX)  # 1D
    # identify unique axis values
    # assume interaction uses columns 1 and 2 in XX (adjust if different)
    x_vals = np.unique(XX[:,1])
    y_vals = np.unique(XX[:,2])
    Z = Z_flat.reshape(len(x_vals), len(y_vals))
    # effect-size metrics on the Z surface
    Z_range = Z.max() - Z.min()
    Z_std = np.std(Z)
    Z_mean_abs = np.mean(np.abs(Z))
    return {'Z_range': Z_range, 'Z_std': Z_std, 'Z_mean_abs': Z_mean_abs}
```

```python
### Check for the interaction of Dst and co2 and compare with interaction of Bx and co2

result_dst_co2 = compute_Z_and_stats(best_gam, 1)
result_dst_co2
```

```python
result_bx_co2 = compute_Z_and_stats(best_gambx, 1)
result_bx_co2
```

### How to compute the stats shown below using a rule of thumb for checking

```python
# Creates a function to compute the normalised range, mean absolute value, and standard deviation
def interpret_interaction_stats(stats, y, ref_stats=None, label='test'):
    """
    stats: dict with 'Z_range', 'Z_std', 'Z_mean_abs' for the interaction surface
    y: observed target array (TEC)
    ref_stats: optional dict for reference strong interaction (e.g., DST×CO2)
    label: string label for printing
    Returns a dict with normalized metrics and textual judgement.
    """
    y_range = np.max(y) - np.min(y)
    y_std = np.std(y)

    Z_range = stats['Z_range']
    Z_mean_abs = stats['Z_mean_abs']
    Z_std = stats['Z_std']

    # Normalized metrics
    R_norm = Z_range / y_range if y_range != 0 else np.nan
    M_norm = Z_mean_abs / y_std if y_std != 0 else np.nan

    # Relative to reference if provided
    ratio_to_ref = None
    if ref_stats is not None and ref_stats.get('Z_range', None) not in (None, 0):
        ratio_to_ref = Z_range / ref_stats['Z_range']

    # Interpret using thresholds above
    def size_label_from_R(R):
        if np.isnan(R): return 'unknown'
        if R < 0.05: return 'negligible'
        if R < 0.15: return 'small'
        if R < 0.35: return 'moderate'
        return 'large'

    def size_label_from_M(M):
        if np.isnan(M): return 'unknown'
        if M < 0.10: return 'negligible'
        if M < 0.30: return 'small'
        if M < 0.60: return 'moderate'
        return 'large'

    if ratio_to_ref is not None:
        if ratio_to_ref < 0.1:
            rel_label = 'tiny vs ref'
        elif ratio_to_ref < 0.3:
            rel_label = 'small vs ref'
        else:
            rel_label = 'material vs ref'
    else:
        rel_label = 'no-ref'

    out = {
        'label': label,
        'Z_range': Z_range,
        'Z_std': Z_std,
        'Z_mean_abs': Z_mean_abs,
        'y_range': y_range,
        'y_std': y_std,
        'R_norm_range': R_norm,
        'M_norm_meanabs': M_norm,
        'ratio_to_ref': ratio_to_ref,
        'range_size_label': size_label_from_R(R_norm),
        'meanabs_size_label': size_label_from_M(M_norm),
        'relative_label': rel_label
    }

    # Nicely print a short summary
    print(f"---- {label} interaction summary ----")
    print(f"Z_range = {Z_range:.3f}, Z_std = {Z_std:.3f}, Z_mean_abs = {Z_mean_abs:.3f}")
    print(f"y_range = {y_range:.3f}, y_std = {y_std:.3f}")
    print(f"Normalized range R_norm = {R_norm:.3f} => {out['range_size_label']}")
    print(f"Normalized mean-abs M_norm = {M_norm:.3f} => {out['meanabs_size_label']}")
    if ratio_to_ref is not None:
        print(f"Ratio to reference Z_range = {ratio_to_ref:.3f} => {rel_label}")
    print("-------------------------------------\n")
    return out
```

```python
# use the function for dst and co2
interpret_interaction_stats(result_dst_co2, Z, label='Dst x Co2 interaction')
```

```python
# use the function for Bx and co2
interpret_interaction_stats(result_bx_co2, Zbx, label='BX x Co2 interaction')
```

#### Plot partial dependence plots

```python
for i, term in enumerate(gamF10_7.terms):
    if term.isintercept:
        continue

    XX = gamF10_7.generate_X_grid(term = i)
    pdep, confi = gamF10_7.partial_dependence(term = i, X = XX, width = 0.95)

    
    
    plt.figure()
    plt.plot(XX[:, term.feature], pdep, label='Partial Dependence', color='blue')
    plt.plot(XX[:, term.feature], confi[:, 0], 'r--', label='Lower 95% CI')
    plt.plot(XX[:, term.feature], confi[:, 1], 'r--', label='Upper 95% CI')
    plt.fill_between(XX[:, term.feature], confi[:, 0], confi[:, 1], color='red', alpha=0.2)  # optional: shaded area
    plt.title(repr(term))
    plt.legend()
    plt.show()
```
